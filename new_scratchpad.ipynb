```python
import dynamiqs as dq
import numpy as np
```

CHECKING DYNAMIQS TENSOR PRODUCT

```python
a = dq.destroy(5)
a_dag = a.dag()
#print(a, a_dag)
I_example = dq.eye(5)
#print(I_example)
full_a = dq.tensor(a, I_example)
full_a_dag = dq.tensor(a_dag, I_example)
n =  full_a + full_a_dag
full_n = dq.tensor((a+a_dag), I_example)
full_n = full_n.to_numpy()
n = n.to_numpy()
check = np.allclose(full_n, n)
check
```
