```python
import numpy as np
import dynamiqs as dq
import qutip as qt
import matplotlib.pyplot as plt
import jax.numpy as jnp
import scipy.constants as const
```

```python
h_bar = const.hbar
omega_q = 6 * 1e-3
Hamiltonian = omega_q *(qt.create(20) * qt.destroy(20))
eval , evec = Hamiltonian.eigenstates()
```

```python
def H_drive(t: float, coupling = 1e6):
    g = 0.5

    drive =  g * np.cos(omega_q * t)
    
    return drive
```

```python
drive_ham = evec[0] * evec[1].dag() + evec[1] * evec[0].dag()

H_driven =  [Hamiltonian, [drive_ham, H_drive]]
```

```python
t = np.arange(0, ((2 * np.pi)/omega_q) , 0.1)
psi_0 = qt.basis(20, 0)
e_ops = [evec[0] * evec[0].dag() ]
solver_options = { "store_states": True}
x = qt.sesolve(H_driven, psi_0, t, e_ops = e_ops, options = solver_options)
```

```python
x.expect
```

```python
T = ((2 * np.pi)/omega_q)
tsave = np.linspace(0, T, 1)
floquet = dq.floquet(H_driven, T, tsave=tsave) 
```
