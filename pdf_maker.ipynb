```python
from pathlib import Path
from pypdf import PdfWriter
import re

folder = Path("/mnt/d/GPU_fluxonium")

output_file = folder/ "Floquet_1000_all_freq_nue_.pdf"

files  = list(folder.glob("Floquet_Branch_Analysis_nue_*_GHz.pdf"))

def get_frequency(path):
    match = re.search(r"phi_([0-9.]+)_GHz\.pdf$", path.name)
    if match:
        return float(match.group(1))
    return float("inf")

files.sort(key = get_frequency)

print(f"Found {len(files)}")

writer = PdfWriter()

for i, pdf_file in enumerate(files):
    freq = get_frequency(pdf_file)
    print(f"{i + 1: 4d}: {freq : .6f} GHz")

    writer.append(str(pdf_file))

with open(output_file, "wb") as f:
    writer.write(f)

print("\nDone")
print(f"Combined File: {output_file}")
```
